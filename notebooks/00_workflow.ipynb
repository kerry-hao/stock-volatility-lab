# Optiver realized-volatility workflow

Read this first. This repository is prepared for learning; no competition data has been downloaded and no new model score has been produced. All notebooks are unexecuted.

**Question:** given 10 minutes of book and trade observations, how much will the stock move during the following 10 minutes? We predict movement magnitude, not direction.


## The unit of prediction

One row in `train.csv` represents one `stock_id` and `time_id` pair. Many raw book/trade records become one feature row. The observed 600 seconds supply inputs; a different subsequent window supplies `target`.

| Stage | Input | Output |
|---|---|---|
| Read | CSV keys and stock-partitioned Parquet | Ordered book/trade records |
| Engineer | WAP, returns, quotes and trades | One feature vector per stock/window |
| Split | Shared `time_id` groups | Disjoint training/validation window groups |
| Fit | Three baselines and LightGBM | Fold-specific models and OOF predictions |
| Evaluate | Held-out targets and predictions | RMSPE comparisons |
| Infer | Matching test features | `row_id,target` CSV |


## Reading order

1. Notebook 01 introduces WAP, returns, RV and persistence using a tiny illustrative sequence.
2. Notebook 02 shows how one local stock becomes a feature table. Its data cells require files you add later.
3. Notebook 03 runs grouped validation and explains the saved results. Training is gated behind a flag.

The implementation lives in `src/optiver_volatility`. The CLI calls those same functions; notebooks do not maintain a second modeling implementation.


## What the supplied reference establishes

The uploaded reference is an introductory Optiver notebook. Its historical persistence result is training RMSPE **0.341** (R² **0.628**) and its text reports public-leaderboard RMSPE **0.327**. These are source results, not this repository's grouped OOF results.

New results remain pending. Read `docs/RESULTS.md` before describing performance publicly.

Source: https://www.kaggle.com/code/jiashenliu/introduction-to-financial-concepts-and-data


## Three decisions to understand

- Historical RV describes the observed window; `target` describes the next one.
- `time_id` groups rows, but its numeric order is not verified chronology.
- Low prediction error is not evidence of a profitable trading strategy.

No download, execution or installation is needed simply to read these notebooks on GitHub.
